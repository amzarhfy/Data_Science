# ***my first data science project***

```python
import pandas as pd #library for data frame
import numpy as np #library/package for dealing with arrays/numbers
```

1. Create a numpy array of size 2x2

```python
arr=np.array(([5,7],[0,9],))
print(arr)
```

2. Create a random numpy array of size 3x4

```python
rand_arr=np.random.rand(3,4)
print(rand_arr)
```

3. Create a numpy array of ones and a numpy array of zeroes

```python
ones=np.ones((5,5))
zeros=np.zeros((3,6))
print(ones)
print(zeros)
```

4. Create a random numpy array of size 6x3 and transpose it

```python
arr=np.random.rand(6,3)
print(arr)
print("transposed array: ")
print(arr.T)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n=2
m=3
p=4

rand_arr1=np.random.rand(n,p)
rand_arr2=np.random.rand(p,m)
print("Array 1: ")
print(rand_arr1)
print("Array 2: ")
print(rand_arr2)
print("Dot product: ")
print(np.dot(rand_arr1,rand_arr2))
```
